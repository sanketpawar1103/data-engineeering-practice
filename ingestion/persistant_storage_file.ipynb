```python
import json

current_policy1 = [{
    "PolicyId": 1,
    "Customer" : "John Doe",
    "Premium" : 1000
}, {
    "PolicyId": 1,
    "Customer" : "John Doe",
    "Premium" : 2000
}];

currentPolicyFile = open("current_policies.json", "a");
currentPolicyFile.write(json.dumps(current_policy1, indent=2))
currentPolicyFile.close()
```
