```python
import json

current_policy1 = [{
    "PolicyId": 1,
    "Customer" : "John Doe",
    "Premium" : 1000
}, {
    "PolicyId": 2,
    "Customer" : "John Doe",
    "Premium" : 2000
}];

currentPolicyFile = open("current_policies.json", "w");
currentPolicyFile.write(json.dumps(current_policy1, indent=2))
currentPolicyFile.close()
```

```python
import json

claims = [{
    "ClaimId": 1,
    "PolicyId": 1,
    "ClaimAmount" : 500
}, {
    "ClaimId": 2,
    "PolicyId": 2,
    "ClaimAmount" : 1000
}];

claimFile = open("claim.json", "w");
claimFile.write(json.dumps(claims, indent=2));
claimFile.close();
```

```python
import json

claimsFile = open("claim.json", "r");
currentPolicyFile = open("current_policies.json", "r");
claimsFileJson = json.load(claimsFile);
currentPoliciesFileJson = json.load(currentPolicyFile);

for policy in currentPoliciesFileJson:
    for claim in claimsFileJson:
        if policy["PolicyId"] == claim["PolicyId"]:
            print(
                f"Policy {policy['PolicyId']} "
                f"-> Premium : {policy['Premium']} "
                f"-> Claim : {claim['ClaimAmount']}"
            )
```
